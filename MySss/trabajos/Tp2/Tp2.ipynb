```python
#PUNTO 1 - FUNCION ESCALON
import numpy as np
import matplotlib.pyplot as plt

def funcion_escalon(t, A, t_o):
  mu = np.where(t >= t_o, 1, 0)

  x = A * mu
  return x

A = 5
t_o = -1
t = np.linspace(-2, 5, 50)

x = funcion_escalon(t, A, t_o)

plt.figure(figsize=(8, 4))
plt.plot(t, x, label=f'$x(t) = {A} \cdot \mu(t - {t_o})$', color='b', linewidth=2)


plt.title('Función Escalón Desplazada')
plt.xlabel('Tiempo (t)')
plt.ylabel('Amplitud x(t)')
plt.axhline(0, color='black', linewidth=1.2) # Eje X
plt.axvline(0, color='black', linewidth=1.2) # Eje Y
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.ylim(-0.5, A + 0.5)


plt.show()
```

```python
# PUNTO 2 - FUNCION SENO

import numpy as np
import matplotlib.pyplot as plt

def funcion_seno(A,f_o,t,theta):
    x = A * np.sin(2 * np.pi * f_o * t + theta)
    return x


A = 1
f_o = 1
t = np.linspace(-2, 5, 100)
theta = 1

x = funcion_seno(A,f_o,t,theta)


plt.figure(figsize=(8, 4))
plt.plot(t, x, label=f'$x(t) = {A} \cdot sen(2\pi \cdot {f_o} \cdot t + {theta})$', color='b', linewidth=2)


plt.title('Función Seno')
plt.xlabel('Tiempo (t)')
plt.ylabel('Amplitud x(t)')
plt.axhline(0, color='black', linewidth=1.2) # Eje X
plt.axvline(0, color='black', linewidth=1.2) # Eje Y
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.ylim(-A - 1, A + 1)


plt.show()
```

```python
# PUNTO 3 - FUNCION EXPONENCIAL

def funcion_exponencial(A,t,t_o,alfa):
    x = A * np.exp(alfa * (t - t_o))
    return x

A = 5
t = np.linspace(-2, 5, 50)
t_o = 1
alfa = 1

x = funcion_exponencial(A,t,t_o,alfa)

plt.figure(figsize=(8, 4))
plt.plot(t, x, label=f'$x(t) = {A} \cdot e^{{{alfa}(t - {t_o})}}$', color='b', linewidth=2)


plt.title('Función Exponencial')
plt.xlabel('Tiempo (t)')
plt.ylabel('Amplitud x(t)')
plt.axhline(0, color='black', linewidth=1.2) # Eje X
plt.axvline(0, color='black', linewidth=1.2) # Eje Y
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.ylim(-A - 1, A + 1)


plt.show()
```

```python
# PUNTO 4 - FUNCION SEÑAL AMORTIGUADA

def funcion_señal_amortigua(A,f_o,t,alfa):
  x = A * np.sin(2 * np.pi * f_o * t + alfa) * np.exp(alfa*t)
  return x

A = 3
f_o = 2
t = np.linspace(-2, 5, 100)
alfa = 1

x = funcion_señal_amortigua(A,f_o,t,alfa)

plt.figure(figsize=(8, 4))
plt.plot(t, x, label=f'$x(t) = {A} \cdot e^{{{alfa}(t - {t_o})}}$', color='b', linewidth=2)


plt.title('Función Señal Amortiguada')
plt.xlabel('Tiempo (t)')
plt.ylabel('Amplitud x(t)')
plt.axhline(0, color='black', linewidth=1.2) # Eje X
plt.axvline(0, color='black', linewidth=1.2) # Eje Y
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show() 
```

```python
# PUNTO 5 - FUNCION PULSO

def funcion_pulso(A,t,D,t_o):
  pulso1 = np.where(t >= t_o, 1, 0)
  pulso2 = np.where(t >= (t_o + D), 1, 0)
  x = A * (pulso1 - pulso2)
  return x

A = 5
t = np.linspace(-2, 5, 100)
D = 2
t_o = -2

x = funcion_pulso(A,t,D,t_o)


plt.figure(figsize=(8, 4))
plt.plot(t, x, label=f'$x(t) = {A} \cdot [\mu(t - {t_o}) - \mu(t - {t_o} - {D})]$', color='b', linewidth=2)


plt.title('Función Pulso')
plt.xlabel('Tiempo (t)')
plt.ylabel('Amplitud x(t)')
plt.axhline(0, color='black', linewidth=1.2) # Eje X
plt.axvline(0, color='black', linewidth=1.2) # Eje Y
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.ylim(-0.5, A + 0.5)

plt.show()
```
